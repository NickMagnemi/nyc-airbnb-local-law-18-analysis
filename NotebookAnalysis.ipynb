```python
import duckdb

PRE_LL18_AIRBNB_DATA: str = 'raw_data/2023-03-06_kaggle_listings_summary.csv'
POST_LL18_AIRBNB_DATA: str = 'raw_data/2026-09-14_listings_detailed.csv.gz'v

duckdb.sql(f"""
    WITH pre_LL18 AS (
        SELECT
        MAX(last_review) AS latest_review
        FROM '{PRE_LL18_AIRBNB_DATA}'
    ),
    post_LL18 AS (
        SELECT
        MAX(last_review) AS latest_review
        FROM '{POST_LL18_AIRBNB_DATA}'
    )
    SELECT * FROM pre_LL18
    UNION ALL BY NAME
    SELECT * FROM post_LL18
""")
```
